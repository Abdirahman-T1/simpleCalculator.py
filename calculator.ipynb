```python
choice = input("do you want to add, subtract, divide or multiply: " +
              "\n" + "use this for add: +" +
               "\n" + "use this for subtract: -" +
              "\n" + "use this for divide: /"+
               "\n" + "use this for multiply: *" + "\n")
firstNum = float(input("first number: "))
secondNum = float(input("second number: "))
if (choice == "+"):
  result = firstNum + secondNum
elif (choice == "-"):
  result = firstNum - secondNum
elif (choice == "*"):
  result = firstNum * secondNum
elif (choice == "/"):
  if (secondNum != 0):
    result = firstNum / secondNum
  else:
    result = "error division by zero"
else:
  print("invalid operation")
print("The answer to your question is: ",result)

```
